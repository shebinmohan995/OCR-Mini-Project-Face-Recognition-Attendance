```python
import cv2
import numpy as np
import os
import face_recognition as fr
from datetime import datetime
```

```python
img_path=r'C:\Users\My Lap\OneDrive\Documents\MLWORK2\DL\ANN\Attendance\Walter white.jpg'
classname=[]
classname.append(os.path.basename(img_path).split('.')[0])
img_path=cv2.imread(img_path)
img=cv2.cvtColor(img_path,cv2.COLOR_BGR2RGB)
known_loc=fr.face_locations(img)
known_code=fr.face_encodings(img)[0]
```

```python
video=cv2.VideoCapture(0)
attendance_marked=False
attendance_time=""
while True:
    ret,frame=video.read()
    if not ret:
        break

    img1=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
    face_loc=fr.face_locations(img1)
    face_code=fr.face_encodings(img1,face_loc)

    for loc,encode in zip(face_loc,face_code):
        face_dis=fr.face_distance([known_code],encode)
        match=fr.compare_faces([known_code],encode)
        face_dis=np.array(face_dis).flatten()
        match=np.array(match).flatten()
        index=np.argmin(face_dis)
        if match[0]==True and face_dis[0]<0.45:
          name=classname[0]
          color=(0,255,0)
          print(name)
          if attendance_marked==False:
              attendance_marked=True
              attendance_time=datetime.now().strftime("%d-%m-%Y %H:%M:%S")
              print("Attendance Marked",name)
              print('Date & Time',attendance_time)
              status="Attandance Marked"
        
        else:
            name="Unknown"
            color=(0,0,255)
            status="Not Recognized"


        y1,x2,y2,x1=loc
        cv2.rectangle(frame,(x1,y1),(x2,y2),color,3)
        cv2.putText(frame,name,(x1,y1),cv2.FONT_HERSHEY_COMPLEX,1,color,2)
        if attendance_marked:
            cv2.putText(frame,attendance_time,(20,40),cv2.FONT_HERSHEY_COMPLEX,0.7,(0,255,0),2)
    cv2.imshow('Attendance System',frame)
    if cv2.waitKey(1)&0XFF==ord('q'):
        break
video.release()
cv2.destroyAllWindows()



```
